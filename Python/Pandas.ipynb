Series and Dataframes are the objects pandas package basically works with 

Series is a 1D array which is labeled and can hold any data type

just like a column in a spreadsheet


```python
data_test= [1, 2, 3, 4, 5]
series = pd.Series(data_test, index=['a', 'b', 'c', 'd', 'e'])
print(series.loc['a'])
```

loc is locaton by label 

What it accepts -	Row/Column names (strings or integers) 

Slicing behavior	Inclusive of the end bound (e.g., loc[1:3] includes 3)

iloc is location by integer

Integer positions only (0 to length-1)

Exclusive of the end bound (e.g., iloc[1:3] stops at 2)

both use the subscript operator "[ ]"

```python
print(series.iloc[0])
```

Finding the value in the series that is greater than 2

```python
print(series[series > 2])
```

A dictionary if is passed into series constructor, it will use the key of dictionary as index automatically

```python
calories= {"day1": 420, "day2": 380, "day3": 390}
myvar = pd.Series(calories)
print(myvar)
```

Dataframe - 2d tabular data structure 

like a full excel spreadhseet 

```python
data_2d= {"Name": ['Alice', 'Bob', 'Charlie'], "Age": [25, 30, 35], "City": ['New York', 'Los Angeles', 'Chicago']}
df = pd.DataFrame(data_2d)
print(df)
```

```python
data_2d= {"Name": ['Alice', 'Bob', 'Charlie'], "Age": [25, 30, 35], "City": ['New York', 'Los Angeles', 'Chicago']}
df = pd.DataFrame(data_2d, index=['row1', 'row2', 'row3'])
print(df)
```

```python
#new row? we just add a new row by making another df and concatenating
#a list of of dictionaries will be used and the key values will be column names and values will be the the new values to be added.

new_row =pd.DataFrame([{"Name": "David", "Age": 40, "City": "Houston"}, {"Name": "Eva", "Age": 28, "City": "Seattle"}]) 
df1=pd.concat([df,new_row],ignore_index=True)
print(df1)
```

```python
#add a new column
df1["Country"] = ["USA", "Canada", "UK", "Australia", "Germany"]
print(df1)
```
